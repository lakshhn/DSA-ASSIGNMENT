## Lab Work 1: Benchmarking Algorithmic Growth Rates

**Objective:**
Empirically investigate constant $O(1)$, logarithmic $O(\log n)$, linear $O(n)$, linearithmic $O(n \log n)$, quadratic $O(n^2)$, and exponential $O(2^n)$ growth patterns.

**Tasks:**
- Implement representative operations/algorithms: list indexing, binary search, linear search, merge sort, pair counter, and subset enumeration.
- Choose safe input-size ranges appropriate to each growth class (small $n$ for exponential growth).
- Exclude data generation from the timed region and use median timing over repeated trials.
- Display timing benchmarks and normalized metrics ($T(n)/\log_2 n$, $T(n)/n$, $T(n)/(n \log_2 n)$, $T(n)/n^2$, $T(n)/2^n$).
- Summarize theoretical bounds vs measured behavior.

```python
# Lakshay - 25BEC056
# Lab 1: Benchmarking Algorithmic Growth Rates

import time
import math
import random


def list_index(arr, idx):
    return arr[idx]


def bin_search(arr, target):
    low = 0
    high = len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1


def lin_search(arr, target):
    for i, val in enumerate(arr):
        if val == target:
            return i
    return -1


def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    
    res = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            res.append(left[i])
            i += 1
        else:
            res.append(right[j])
            j += 1
    res.extend(left[i:])
    res.extend(right[j:])
    return res


def count_pairs(arr):
    count = 0
    n = len(arr)
    for i in range(n):
        for j in range(i + 1, n):
            if arr[i] + arr[j] > 0:
                count += 1
    return count


def subset_enum(arr):
    res = [[]]
    for x in arr:
        res += [sub + [x] for sub in res]
    return res


def median_time(fn, *args, trials=5):
    times = []
    for _ in range(trials):
        t0 = time.perf_counter()
        fn(*args)
        t1 = time.perf_counter()
        times.append(t1 - t0)
    times.sort()
    return times[len(times) // 2]


def run_benchmarks():
    print("1. Constant Growth O(1) - List Indexing")
    for n in [10000, 50000, 100000, 250000, 500000]:
        arr = list(range(n))
        t = median_time(list_index, arr, n // 2)
        print(f"  n = {n:<8} | Time: {t:.8f} s")

    print("\n2. Logarithmic Growth O(log n) - Binary Search")
    for n in [10000, 50000, 100000, 250000, 500000]:
        arr = list(range(n))
        t = median_time(bin_search, arr, -1)
        norm = t / math.log2(n)
        print(f"  n = {n:<8} | Time: {t:.8f} s | T / log2(n): {norm:.2e}")

    print("\n3. Linear Growth O(n) - Linear Search")
    for n in [10000, 50000, 100000, 200000, 400000]:
        arr = list(range(n))
        t = median_time(lin_search, arr, -1)
        norm = t / n
        print(f"  n = {n:<8} | Time: {t:.8f} s | T / n: {norm:.2e}")

    print("\n4. Linearithmic Growth O(n log n) - Merge Sort")
    for n in [1000, 2000, 4000, 8000, 16000]:
        arr = [random.randint(1, 100000) for _ in range(n)]
        t = median_time(merge_sort, arr)
        norm = t / (n * math.log2(n))
        print(f"  n = {n:<8} | Time: {t:.8f} s | T / (n log2 n): {norm:.2e}")

    print("\n5. Quadratic Growth O(n^2) - Pair Counter")
    for n in [200, 400, 800, 1600, 3200]:
        arr = [random.randint(-100, 100) for _ in range(n)]
        t = median_time(count_pairs, arr)
        norm = t / (n ** 2)
        print(f"  n = {n:<8} | Time: {t:.8f} s | T / n^2: {norm:.2e}")

    print("\n6. Exponential Growth O(2^n) - Subset Enumeration")
    for n in [10, 12, 14, 16, 18, 20]:
        arr = list(range(n))
        t = median_time(subset_enum, arr, trials=3)
        norm = t / (2 ** n)
        print(f"  n = {n:<8} | Time: {t:.8f} s | T / 2^n: {norm:.2e}")

    print("\nSummary:")
    summary = [
        ("List Indexing", "O(1)", "Flat lookup"),
        ("Binary Search", "O(log n)", "Very slow growth"),
        ("Linear Search", "O(n)", "T(n) grows with n"),
        ("Merge Sort", "O(n log n)", "Efficient divide & conquer"),
        ("Pair Counter", "O(n^2)", "4x time when n doubles"),
        ("Subset Enum", "O(2^n)", "Doubles with each +1"),
    ]
    for name, bound, desc in summary:
        print(f"  {name:<16} | {bound:<10} | {desc}")


if __name__ == "__main__":
    run_benchmarks()
```

## Lab Work 2: Prefix Averages: Quadratic versus Linear-Time Design

**Objective:**
Demonstrate the reduction from quadratic to linear time using a running-sum strategy for computing prefix averages.

**Tasks:**
- Implement `prefix_average1` using a nested loop that recomputes each prefix sum ($O(n^2)$ time).
- Implement `prefix_average2` using a running sum updated once per element ($O(n)$ time).
- Verify equivalent outputs across diverse test sequences (empty, singleton, positive, negative, and mixed values).
- Benchmark both implementations across increasing sequence lengths ($n = 500$ to $8000$) and compute the speedup factor.

```python
# Lakshay - 25BEC056
# Lab 2: Prefix Averages (Quadratic vs Linear Time)

import time
import math
import random


def prefix_average1(arr):
    n = len(arr)
    res = [0.0] * n
    for i in range(n):
        total = 0.0
        for j in range(i + 1):
            total += arr[j]
        res[i] = total / (i + 1)
    return res


def prefix_average2(arr):
    n = len(arr)
    res = [0.0] * n
    total = 0.0
    for i in range(n):
        total += arr[i]
        res[i] = total / (i + 1)
    return res


def median_time(fn, arr, trials=5):
    times = []
    for _ in range(trials):
        t0 = time.perf_counter()
        fn(arr)
        t1 = time.perf_counter()
        times.append(t1 - t0)
    times.sort()
    return times[len(times) // 2]


def validate():
    test_cases = [
        ("Empty", []),
        ("Singleton", [42.0]),
        ("Positive", [1, 2, 3, 4, 5]),
        ("Negative", [-2, -4, -6, -8]),
        ("Mixed", [10, -5, 20, -10, 50]),
    ]
    print("Validation:")
    for desc, seq in test_cases:
        r1 = prefix_average1(seq)
        r2 = prefix_average2(seq)
        match = len(r1) == len(r2) and all(
            math.isclose(a, b, rel_tol=1e-9) for a, b in zip(r1, r2)
        )
        status = "Passed" if match else "Failed"
        print(f"  {desc:<12}: {status} -> {[round(x, 2) for x in r2]}")


def benchmark():
    print("\nBenchmarking (prefix_average1 vs prefix_average2):")
    sizes = [500, 1000, 2000, 4000, 8000]
    for n in sizes:
        arr = [random.uniform(1.0, 100.0) for _ in range(n)]
        t1 = median_time(prefix_average1, arr)
        t2 = median_time(prefix_average2, arr)
        speedup = t1 / t2 if t2 > 0 else 0
        print(f"  n = {n:<5} | O(n^2): {t1:.6f} s | O(n): {t2:.6f} s | Speedup: {speedup:.1f}x")


if __name__ == "__main__":
    validate()
    benchmark()
```
