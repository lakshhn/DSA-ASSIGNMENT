### Question 3: Singly Linked List Operations
Develop algorithms for `insert_first`, `insert_last`, `delete_first`, and traversal for a singly linked list. State the preconditions and pointer updates for each.

```python
# Lakshay - 25BEC056
# Part D - Question 3: Singly Linked List operations (insert_first, insert_last, delete_first, traversal)


class Node:
    def __init__(self, val, next_node=None):
        self.val = val
        self.next = next_node


class SinglyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    def is_empty(self):
        return self.size == 0

    def insert_first(self, val):
        new_node = Node(val, self.head)
        self.head = new_node
        if self.size == 0:
            self.tail = new_node
        self.size += 1

    def insert_last(self, val):
        new_node = Node(val)
        if self.is_empty():
            self.head = new_node
        else:
            self.tail.next = new_node
        self.tail = new_node
        self.size += 1

    def delete_first(self):
        if self.is_empty():
            raise IndexError("Cannot delete from an empty list")
        val = self.head.val
        self.head = self.head.next
        self.size -= 1
        if self.is_empty():
            self.tail = None
        return val

    def traverse(self):
        curr = self.head
        elements = []
        while curr:
            elements.append(curr.val)
            curr = curr.next
        return elements


if __name__ == "__main__":
    sll = SinglyLinkedList()
    sll.insert_first(20)
    sll.insert_first(10)
    sll.insert_last(30)
    sll.insert_last(40)
    print("List after insertions:", sll.traverse())
    print("Deleted first:", sll.delete_first())
    print("List after delete_first:", sll.traverse())
```

### Question 6: Favorites List ADT
Design a Favorites List ADT supporting `access(e)`, `remove(e)`, and `top(k)`. Specify the stored item fields and update behaviour.

```python
# Lakshay - 25BEC056
# Part D - Question 6: Favorites List ADT


class FavoritesList:
    class _Item:
        def __init__(self, val):
            self.val = val
            self.count = 0

    def __init__(self):
        self.data = []

    def is_empty(self):
        return len(self.data) == 0

    def access(self, e):
        for item in self.data:
            if item.val == e:
                item.count += 1
                return
        new_item = self._Item(e)
        new_item.count = 1
        self.data.append(new_item)

    def remove(self, e):
        for i, item in enumerate(self.data):
            if item.val == e:
                self.data.pop(i)
                return True
        return False

    def top(self, k):
        sorted_items = sorted(self.data, key=lambda x: x.count, reverse=True)
        return [(item.val, item.count) for item in sorted_items[:k]]


if __name__ == "__main__":
    fav = FavoritesList()
    for page in ["google.com", "github.com", "google.com", "stackoverflow.com", "github.com", "google.com"]:
        fav.access(page)

    print("Top 2 favorites:")
    for url, count in fav.top(2):
        print(f"  {url} ({count} accesses)")

    fav.remove("github.com")
    print("\nAfter removing github.com:")
    for url, count in fav.top(2):
        print(f"  {url} ({count} accesses)")
```
