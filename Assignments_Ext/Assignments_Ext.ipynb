### Question 2: Recursive Factorial, Sum, and Power
Write recursive Python functions for factorial, sum of the first $n$ natural numbers, and computation of $x$ raised to a non-negative integer power. State preconditions and base cases.

```python
# Lakshay - 25BEC056
# Part A - Question 2: Recursive factorial, sum, and power


def fact(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n == 0:
        return 1
    return n * fact(n - 1)


def sum_n(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n == 0:
        return 0
    return n + sum_n(n - 1)


def power(x, n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n == 0:
        return 1
    return x * power(x, n - 1)


if __name__ == "__main__":
    print("Factorial of 5:", fact(5))
    print("Sum of first 10 natural numbers:", sum_n(10))
    print("2 raised to power 8:", power(2, 8))
```

### Question 3: In-Place Sequence Reversal
Design a recursive algorithm to reverse a sequence in place using start and stop indices. Trace its operation for a sequence of six elements.

```python
# Lakshay - 25BEC056
# Part A - Question 3: In-place sequence reversal using start and stop indices


def reverse_inplace(arr, start, stop):
    if start >= stop:
        return

    print(f"Swap arr[{start}]={arr[start]} with arr[{stop}]={arr[stop]} -> state: {arr}")
    arr[start], arr[stop] = arr[stop], arr[start]

    reverse_inplace(arr, start + 1, stop - 1)


if __name__ == "__main__":
    nums = [10, 20, 30, 40, 50, 60]
    print(f"Original sequence: {nums}")
    print("\nTrace of recursive swaps:")
    reverse_inplace(nums, 0, len(nums) - 1)
    print(f"\nFinal reversed sequence: {nums}")
```

### Question 4: Recursive Sorted Sequence Check
Write a recursive function that checks whether a sequence is sorted in non-decreasing order. State its best- and worst-case running time.

```python
# Lakshay - 25BEC056
# Part A - Question 4: Recursive check if a sequence is sorted non-decreasingly


def is_sorted(arr, idx=0):
    if len(arr) <= 1 or idx >= len(arr) - 1:
        return True

    if arr[idx] > arr[idx + 1]:
        return False

    return is_sorted(arr, idx + 1)


if __name__ == "__main__":
    seq1 = [2, 5, 8, 12, 19, 25]
    seq2 = [2, 5, 15, 8, 19, 25]
    seq3 = [10]
    seq4 = []

    print(f"{seq1} -> sorted: {is_sorted(seq1)}")
    print(f"{seq2} -> sorted: {is_sorted(seq2)}")
    print(f"{seq3} -> sorted: {is_sorted(seq3)}")
    print(f"{seq4} -> sorted: {is_sorted(seq4)}")
```

### Question 5: Naive Recursive Fibonacci and Recursion Tree
Write the naive recursive Fibonacci algorithm and draw its recursion tree for $n=5$. Identify repeated subproblems and count recursive calls.

```python
# Lakshay - 25BEC056
# Part A - Question 5: Naive recursive Fibonacci and call counts for n=5

call_counts = {}


def fib(n):
    call_counts[n] = call_counts.get(n, 0) + 1

    if n <= 0:
        return 0
    elif n == 1:
        return 1

    return fib(n - 1) + fib(n - 2)


def print_recursion_tree_for_5():
    tree = r"""
Recursion Tree for fib(5):

                       fib(5)
                     /        \
              fib(4)            fib(3)
             /      \          /      \
        fib(3)      fib(2)    fib(2)  fib(1)
       /     \      /    \    /    \
    fib(2) fib(1) fib(1) fib(0) fib(1) fib(0)
    /    \
 fib(1) fib(0)
"""
    print(tree)


if __name__ == "__main__":
    call_counts.clear()
    val = fib(5)

    print(f"fib(5) = {val}")
    print_recursion_tree_for_5()

    print("Call counts per subproblem:")
    for arg in sorted(call_counts.keys(), reverse=True):
        print(f"  fib({arg}) called {call_counts[arg]} time(s)")
```

### Question 7: Tower of Hanoi
Develop a recursive algorithm for the Tower of Hanoi problem for $n$ disks. State the three recursive steps and the base case.

```python
# Lakshay - 25BEC056
# Part A - Question 7: Recursive Tower of Hanoi for n disks


def hanoi(n, src, dst, aux, step=[0]):
    if n == 1:
        step[0] += 1
        print(f"Step {step[0]:>2}: Move disk 1 from {src} -> {dst}")
        return

    hanoi(n - 1, src, aux, dst, step)

    step[0] += 1
    print(f"Step {step[0]:>2}: Move disk {n} from {src} -> {dst}")

    hanoi(n - 1, aux, dst, src, step)


if __name__ == "__main__":
    n_disks = 3
    print(f"Tower of Hanoi solution for n = {n_disks} disks:")
    hanoi(n_disks, "A", "C", "B")
```

### Question 9: Tail Recursion to Iterative Conversion
Convert a tail-recursive sequence-processing algorithm into an iterative version. Compare the use of call-stack space in both versions.

```python
# Lakshay - 25BEC056
# Part A - Question 9: Converting tail-recursive sum to iterative version


def sum_tail(arr, idx=0, acc=0):
    if idx >= len(arr):
        return acc
    return sum_tail(arr, idx + 1, acc + arr[idx])


def sum_iter(arr):
    acc = 0
    for val in arr:
        acc += val
    return acc


if __name__ == "__main__":
    sample = [5, 10, 15, 20, 25]
    print(f"Input: {sample}")
    print("Tail-recursive result:", sum_tail(sample))
    print("Iterative result:     ", sum_iter(sample))
```

### Question 2: Stack Implementation using Python List
Implement a Stack class using a Python list. Use a suitable exception for `pop()` and `top()` on an empty stack.

```python
# Lakshay - 25BEC056
# Part C - Question 2: Stack class using a Python list


class EmptyStackError(Exception):
    pass


class Stack:
    def __init__(self):
        self.data = []

    def push(self, val):
        self.data.append(val)

    def pop(self):
        if self.is_empty():
            raise EmptyStackError("Cannot pop from an empty stack")
        return self.data.pop()

    def top(self):
        if self.is_empty():
            raise EmptyStackError("Cannot view top of an empty stack")
        return self.data[-1]

    def is_empty(self):
        return len(self.data) == 0

    def size(self):
        return len(self.data)


if __name__ == "__main__":
    s = Stack()
    print("Stack created. is_empty:", s.is_empty())

    s.push(10)
    s.push(20)
    s.push(30)
    print("Pushed 10, 20, 30. Size:", s.size())
    print("Top:", s.top())

    print("Popped:", s.pop())
    print("New top:", s.top())
```

### Question 8: Circular Array Queue
Design a circular-array queue using `data`, `size`, and `front` variables. Derive the index formula for the next insertion and the next front position.

```python
# Lakshay - 25BEC056
# Part C - Question 8: Circular array queue


class EmptyQueueError(Exception):
    pass


class CircularQueue:
    def __init__(self, cap=5):
        self.cap = cap
        self.data = [None] * cap
        self.size = 0
        self.front = 0

    def is_empty(self):
        return self.size == 0

    def is_full(self):
        return self.size == self.cap

    def enqueue(self, val):
        if self.is_full():
            raise OverflowError("Queue is full")
        ins_idx = (self.front + self.size) % self.cap
        self.data[ins_idx] = val
        self.size += 1

    def dequeue(self):
        if self.is_empty():
            raise EmptyQueueError("Queue is empty")
        val = self.data[self.front]
        self.data[self.front] = None
        self.front = (self.front + 1) % self.cap
        self.size -= 1
        return val

    def first(self):
        if self.is_empty():
            raise EmptyQueueError("Queue is empty")
        return self.data[self.front]


if __name__ == "__main__":
    q = CircularQueue(cap=5)
    for item in [10, 20, 30, 40]:
        q.enqueue(item)
        print(f"Enqueued {item}")

    print("First:", q.first())
    print("Dequeued:", q.dequeue())
    print("Dequeued:", q.dequeue())

    q.enqueue(50)
    q.enqueue(60)
    print("Enqueued 50, 60")
    print("Current first:", q.first())
```

### Question 10: Circular Array Deque
Develop algorithms for `add_first`, `add_last`, `delete_first`, and `delete_last` using circular-array indexing.

```python
# Lakshay - 25BEC056
# Part C - Question 10: Circular array Deque (Double-Ended Queue)


class EmptyDequeError(Exception):
    pass


class CircularDeque:
    def __init__(self, cap=5):
        self.cap = cap
        self.data = [None] * cap
        self.size = 0
        self.front = 0

    def is_empty(self):
        return self.size == 0

    def is_full(self):
        return self.size == self.cap

    def add_first(self, e):
        if self.is_full():
            raise OverflowError("Deque is full")
        self.front = (self.front - 1) % self.cap
        self.data[self.front] = e
        self.size += 1

    def add_last(self, e):
        if self.is_full():
            raise OverflowError("Deque is full")
        back = (self.front + self.size) % self.cap
        self.data[back] = e
        self.size += 1

    def delete_first(self):
        if self.is_empty():
            raise EmptyDequeError("Deque is empty")
        val = self.data[self.front]
        self.data[self.front] = None
        self.front = (self.front + 1) % self.cap
        self.size -= 1
        return val

    def delete_last(self):
        if self.is_empty():
            raise EmptyDequeError("Deque is empty")
        back = (self.front + self.size - 1) % self.cap
        val = self.data[back]
        self.data[back] = None
        self.size -= 1
        return val


if __name__ == "__main__":
    dq = CircularDeque(cap=5)
    dq.add_last(10)
    dq.add_last(20)
    dq.add_first(5)
    print("Added last 10, 20 and added first 5")
    print("Deleted first:", dq.delete_first())
    print("Deleted last:", dq.delete_last())
```

### Question 3: Singly Linked List Operations
Develop algorithms for `insert_first`, `insert_last`, `delete_first`, and traversal for a singly linked list. State the preconditions and pointer updates for each.

```python
# Lakshay - 25BEC056
# Part D - Question 3: Singly Linked List operations (insert_first, insert_last, delete_first, traversal)


class Node:
    def __init__(self, val, next_node=None):
        self.val = val
        self.next = next_node


class SinglyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    def is_empty(self):
        return self.size == 0

    def insert_first(self, val):
        new_node = Node(val, self.head)
        self.head = new_node
        if self.size == 0:
            self.tail = new_node
        self.size += 1

    def insert_last(self, val):
        new_node = Node(val)
        if self.is_empty():
            self.head = new_node
        else:
            self.tail.next = new_node
        self.tail = new_node
        self.size += 1

    def delete_first(self):
        if self.is_empty():
            raise IndexError("Cannot delete from an empty list")
        val = self.head.val
        self.head = self.head.next
        self.size -= 1
        if self.is_empty():
            self.tail = None
        return val

    def traverse(self):
        curr = self.head
        elements = []
        while curr:
            elements.append(curr.val)
            curr = curr.next
        return elements


if __name__ == "__main__":
    sll = SinglyLinkedList()
    sll.insert_first(20)
    sll.insert_first(10)
    sll.insert_last(30)
    sll.insert_last(40)
    print("List after insertions:", sll.traverse())
    print("Deleted first:", sll.delete_first())
    print("List after delete_first:", sll.traverse())
```

### Question 6: Favorites List ADT
Design a Favorites List ADT supporting `access(e)`, `remove(e)`, and `top(k)`. Specify the stored item fields and update behaviour.

```python
# Lakshay - 25BEC056
# Part D - Question 6: Favorites List ADT


class FavoritesList:
    class _Item:
        def __init__(self, val):
            self.val = val
            self.count = 0

    def __init__(self):
        self.data = []

    def is_empty(self):
        return len(self.data) == 0

    def access(self, e):
        for item in self.data:
            if item.val == e:
                item.count += 1
                return
        new_item = self._Item(e)
        new_item.count = 1
        self.data.append(new_item)

    def remove(self, e):
        for i, item in enumerate(self.data):
            if item.val == e:
                self.data.pop(i)
                return True
        return False

    def top(self, k):
        sorted_items = sorted(self.data, key=lambda x: x.count, reverse=True)
        return [(item.val, item.count) for item in sorted_items[:k]]


if __name__ == "__main__":
    fav = FavoritesList()
    for page in ["google.com", "github.com", "google.com", "stackoverflow.com", "github.com", "google.com"]:
        fav.access(page)

    print("Top 2 favorites:")
    for url, count in fav.top(2):
        print(f"  {url} ({count} accesses)")

    fav.remove("github.com")
    print("\nAfter removing github.com:")
    for url, count in fav.top(2):
        print(f"  {url} ({count} accesses)")
```
