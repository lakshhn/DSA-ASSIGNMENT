### Question 2: Recursive Factorial, Sum, and Power
Write recursive Python functions for factorial, sum of the first $n$ natural numbers, and computation of $x$ raised to a non-negative integer power. State preconditions and base cases.

```python
# Lakshay - 25BEC056
# Part A - Question 2: Recursive factorial, sum, and power


def fact(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n == 0:
        return 1
    return n * fact(n - 1)


def sum_n(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n == 0:
        return 0
    return n + sum_n(n - 1)


def power(x, n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n == 0:
        return 1
    return x * power(x, n - 1)


if __name__ == "__main__":
    print("Factorial of 5:", fact(5))
    print("Sum of first 10 natural numbers:", sum_n(10))
    print("2 raised to power 8:", power(2, 8))
```

### Question 3: In-Place Sequence Reversal
Design a recursive algorithm to reverse a sequence in place using start and stop indices. Trace its operation for a sequence of six elements.

```python
# Lakshay - 25BEC056
# Part A - Question 3: In-place sequence reversal using start and stop indices


def reverse_inplace(arr, start, stop):
    if start >= stop:
        return

    print(f"Swap arr[{start}]={arr[start]} with arr[{stop}]={arr[stop]} -> state: {arr}")
    arr[start], arr[stop] = arr[stop], arr[start]

    reverse_inplace(arr, start + 1, stop - 1)


if __name__ == "__main__":
    nums = [10, 20, 30, 40, 50, 60]
    print(f"Original sequence: {nums}")
    print("\nTrace of recursive swaps:")
    reverse_inplace(nums, 0, len(nums) - 1)
    print(f"\nFinal reversed sequence: {nums}")
```

### Question 4: Recursive Sorted Sequence Check
Write a recursive function that checks whether a sequence is sorted in non-decreasing order. State its best- and worst-case running time.

```python
# Lakshay - 25BEC056
# Part A - Question 4: Recursive check if a sequence is sorted non-decreasingly


def is_sorted(arr, idx=0):
    if len(arr) <= 1 or idx >= len(arr) - 1:
        return True

    if arr[idx] > arr[idx + 1]:
        return False

    return is_sorted(arr, idx + 1)


if __name__ == "__main__":
    seq1 = [2, 5, 8, 12, 19, 25]
    seq2 = [2, 5, 15, 8, 19, 25]
    seq3 = [10]
    seq4 = []

    print(f"{seq1} -> sorted: {is_sorted(seq1)}")
    print(f"{seq2} -> sorted: {is_sorted(seq2)}")
    print(f"{seq3} -> sorted: {is_sorted(seq3)}")
    print(f"{seq4} -> sorted: {is_sorted(seq4)}")
```

### Question 5: Naive Recursive Fibonacci and Recursion Tree
Write the naive recursive Fibonacci algorithm and draw its recursion tree for $n=5$. Identify repeated subproblems and count recursive calls.

```python
# Lakshay - 25BEC056
# Part A - Question 5: Naive recursive Fibonacci and call counts for n=5

call_counts = {}


def fib(n):
    call_counts[n] = call_counts.get(n, 0) + 1

    if n <= 0:
        return 0
    elif n == 1:
        return 1

    return fib(n - 1) + fib(n - 2)


def print_recursion_tree_for_5():
    tree = r"""
Recursion Tree for fib(5):

                       fib(5)
                     /        \
              fib(4)            fib(3)
             /      \          /      \
        fib(3)      fib(2)    fib(2)  fib(1)
       /     \      /    \    /    \
    fib(2) fib(1) fib(1) fib(0) fib(1) fib(0)
    /    \
 fib(1) fib(0)
"""
    print(tree)


if __name__ == "__main__":
    call_counts.clear()
    val = fib(5)

    print(f"fib(5) = {val}")
    print_recursion_tree_for_5()

    print("Call counts per subproblem:")
    for arg in sorted(call_counts.keys(), reverse=True):
        print(f"  fib({arg}) called {call_counts[arg]} time(s)")
```

### Question 7: Tower of Hanoi
Develop a recursive algorithm for the Tower of Hanoi problem for $n$ disks. State the three recursive steps and the base case.

```python
# Lakshay - 25BEC056
# Part A - Question 7: Recursive Tower of Hanoi for n disks


def hanoi(n, src, dst, aux, step=[0]):
    if n == 1:
        step[0] += 1
        print(f"Step {step[0]:>2}: Move disk 1 from {src} -> {dst}")
        return

    hanoi(n - 1, src, aux, dst, step)

    step[0] += 1
    print(f"Step {step[0]:>2}: Move disk {n} from {src} -> {dst}")

    hanoi(n - 1, aux, dst, src, step)


if __name__ == "__main__":
    n_disks = 3
    print(f"Tower of Hanoi solution for n = {n_disks} disks:")
    hanoi(n_disks, "A", "C", "B")
```

### Question 9: Tail Recursion to Iterative Conversion
Convert a tail-recursive sequence-processing algorithm into an iterative version. Compare the use of call-stack space in both versions.

```python
# Lakshay - 25BEC056
# Part A - Question 9: Converting tail-recursive sum to iterative version


def sum_tail(arr, idx=0, acc=0):
    if idx >= len(arr):
        return acc
    return sum_tail(arr, idx + 1, acc + arr[idx])


def sum_iter(arr):
    acc = 0
    for val in arr:
        acc += val
    return acc


if __name__ == "__main__":
    sample = [5, 10, 15, 20, 25]
    print(f"Input: {sample}")
    print("Tail-recursive result:", sum_tail(sample))
    print("Iterative result:     ", sum_iter(sample))
```
