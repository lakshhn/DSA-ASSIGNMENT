### Question 2: Stack Implementation using Python List
Implement a Stack class using a Python list. Use a suitable exception for `pop()` and `top()` on an empty stack.

```python
# Lakshay - 25BEC056
# Part C - Question 2: Stack class using a Python list


class EmptyStackError(Exception):
    pass


class Stack:
    def __init__(self):
        self.data = []

    def push(self, val):
        self.data.append(val)

    def pop(self):
        if self.is_empty():
            raise EmptyStackError("Cannot pop from an empty stack")
        return self.data.pop()

    def top(self):
        if self.is_empty():
            raise EmptyStackError("Cannot view top of an empty stack")
        return self.data[-1]

    def is_empty(self):
        return len(self.data) == 0

    def size(self):
        return len(self.data)


if __name__ == "__main__":
    s = Stack()
    print("Stack created. is_empty:", s.is_empty())

    s.push(10)
    s.push(20)
    s.push(30)
    print("Pushed 10, 20, 30. Size:", s.size())
    print("Top:", s.top())

    print("Popped:", s.pop())
    print("New top:", s.top())
```

### Question 8: Circular Array Queue
Design a circular-array queue using `data`, `size`, and `front` variables. Derive the index formula for the next insertion and the next front position.

```python
# Lakshay - 25BEC056
# Part C - Question 8: Circular array queue


class EmptyQueueError(Exception):
    pass


class CircularQueue:
    def __init__(self, cap=5):
        self.cap = cap
        self.data = [None] * cap
        self.size = 0
        self.front = 0

    def is_empty(self):
        return self.size == 0

    def is_full(self):
        return self.size == self.cap

    def enqueue(self, val):
        if self.is_full():
            raise OverflowError("Queue is full")
        ins_idx = (self.front + self.size) % self.cap
        self.data[ins_idx] = val
        self.size += 1

    def dequeue(self):
        if self.is_empty():
            raise EmptyQueueError("Queue is empty")
        val = self.data[self.front]
        self.data[self.front] = None
        self.front = (self.front + 1) % self.cap
        self.size -= 1
        return val

    def first(self):
        if self.is_empty():
            raise EmptyQueueError("Queue is empty")
        return self.data[self.front]


if __name__ == "__main__":
    q = CircularQueue(cap=5)
    for item in [10, 20, 30, 40]:
        q.enqueue(item)
        print(f"Enqueued {item}")

    print("First:", q.first())
    print("Dequeued:", q.dequeue())
    print("Dequeued:", q.dequeue())

    q.enqueue(50)
    q.enqueue(60)
    print("Enqueued 50, 60")
    print("Current first:", q.first())
```

### Question 10: Circular Array Deque
Develop algorithms for `add_first`, `add_last`, `delete_first`, and `delete_last` using circular-array indexing.

```python
# Lakshay - 25BEC056
# Part C - Question 10: Circular array Deque (Double-Ended Queue)


class EmptyDequeError(Exception):
    pass


class CircularDeque:
    def __init__(self, cap=5):
        self.cap = cap
        self.data = [None] * cap
        self.size = 0
        self.front = 0

    def is_empty(self):
        return self.size == 0

    def is_full(self):
        return self.size == self.cap

    def add_first(self, e):
        if self.is_full():
            raise OverflowError("Deque is full")
        self.front = (self.front - 1) % self.cap
        self.data[self.front] = e
        self.size += 1

    def add_last(self, e):
        if self.is_full():
            raise OverflowError("Deque is full")
        back = (self.front + self.size) % self.cap
        self.data[back] = e
        self.size += 1

    def delete_first(self):
        if self.is_empty():
            raise EmptyDequeError("Deque is empty")
        val = self.data[self.front]
        self.data[self.front] = None
        self.front = (self.front + 1) % self.cap
        self.size -= 1
        return val

    def delete_last(self):
        if self.is_empty():
            raise EmptyDequeError("Deque is empty")
        back = (self.front + self.size - 1) % self.cap
        val = self.data[back]
        self.data[back] = None
        self.size -= 1
        return val


if __name__ == "__main__":
    dq = CircularDeque(cap=5)
    dq.add_last(10)
    dq.add_last(20)
    dq.add_first(5)
    print("Added last 10, 20 and added first 5")
    print("Deleted first:", dq.delete_first())
    print("Deleted last:", dq.delete_last())
```
