### Q3. Strings
Write a Python program to accept a string and: Count the number of characters, Count vowels and consonants, Reverse the string, Check whether it is a palindrome.

```python
s = input("enter a string: ")

print("length:", len(s))

v = 0
c = 0
for ch in s.lower():
    if ch.isalpha():
        if ch in "aeiou":
            v += 1
        else:
            c += 1

print("vowels:", v)
print("consonants:", c)

r = s[::-1]
print("reversed:", r)

# just comparing with reversed one to check palindrome
if s.lower() == r.lower():
    print("its a palindrome")
else:
    print("not a palindrome")
```

### Q5. GPA Calculator
Develop a Python program that accepts the grade/grade point and credit of different courses and calculates the student's SGPA/GPA using: GPA = Σ(Grade Point × Credit) / ΣCredits

```python
n = int(input("how many courses? "))
tp = 0
tc = 0

for i in range(n):
    gp = float(input("grade point: "))
    cr = float(input("credit: "))
    tp += gp * cr
    tc += cr

gpa = tp / tc
print("gpa =", round(gpa, 2))
```

### Q6. Functions and Iteration
Write separate Python functions to: Calculate the factorial of a number, Generate the Fibonacci series up to n terms, Check whether a number is prime.

```python
def fact(n):
    f = 1
    for i in range(1, n + 1):
        f *= i
    return f

def fib(n):
    a, b = 0, 1
    res = []
    for i in range(n):
        res.append(a)
        a, b = b, a + b
    return res

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

print("factorial of 5:", fact(5))
print("fib series:", fib(10))
print("is 17 prime?", is_prime(17))
print("is 20 prime?", is_prime(20))
```

### Q9. List Comprehension
Using list comprehension, generate: Squares of numbers from 1 to 20, Even numbers from 1 to 50, Numbers divisible by both 3 and 5 from 1 to 100.

```python
sq = [i ** 2 for i in range(1, 21)]
print(sq)

ev = [i for i in range(1, 51) if i % 2 == 0]
print(ev)

d = [i for i in range(1, 101) if i % 3 == 0 and i % 5 == 0]
print(d)
```

### Q10. Mini Programming Problem
Develop a menu-driven Python program for maintaining a student record using a dictionary. The program should provide options to: Add a student, Update marks, Search for a student, Delete a student, Display all student records.

```python
rec = {}

while True:
    print("\n1 add  2 update  3 search  4 delete  5 show all  6 exit")
    ch = input("choice: ")

    if ch == "1":
        nm = input("name: ")
        mk = float(input("marks: "))
        rec[nm] = mk
        print("added")

    elif ch == "2":
        nm = input("name: ")
        if nm in rec:
            mk = float(input("new marks: "))
            rec[nm] = mk
            print("updated")
        else:
            print("not found")

    elif ch == "3":
        nm = input("name: ")
        if nm in rec:
            print(nm, "-", rec[nm])
        else:
            print("not found")

    elif ch == "4":
        nm = input("name: ")
        if nm in rec:
            del rec[nm]
            print("deleted")
        else:
            print("not found")

    elif ch == "5":
        if len(rec) == 0:
            print("no records yet")
        else:
            for nm in rec:
                print(nm, "-", rec[nm])

    elif ch == "6":
        break

    else:
        print("wrong choice, try again")
```
